```python
PLMD = {
    'P11498':[741],
    'P27708':[1556],
    'P05020':[103],
    'P0A6B4':[122],
    'P13661':[74],
    'P22188':[225],
    'P39377':[162],
    'P0C512':[201],
    'O93627':[189],
    'P00875':[201],
    'P00876':[201],
    'P00877':[201],
    'P00880':[198],
    'P04718':[191],
    'P07374':[490],
    'P0A1V8':[75],
    'P0A2W8':[129],
    'P0A433':[169],
    'P0A434':[169],
    'P0C2C2':[204],
    'P10724':[129],
    'P14489':[70],
    'P18314':[217],
    'P41020':[219],
    'P69996':[219],
    'P81006':[147],
    'Q45515':[150],
    'Q55DL0':[158],
    'Q5L1E2':[173],
    'Q70AC7':[184],
    'Q7CS13':[173],
    'Q837K0':[152],
    'Q8VTT5':[148],
    'Q97VT7':[137],
    'Q9HTQ2':[122],
    'Q9P903':[167],
}
```

```python
UP = Uniprot()
for entry in PLMD.keys():
    UP.get_protein_data(entry)
```

```python
UP.df
```

```python
check = UP.df
check['Label'] = ['Included'] * len(check)
```

```python
pdb = PDB(outdir = outdir, gap = 10)
```

```python
list_of_metals = ['ZN', 'NI', 'CU', 'FE', 'MG', 'MN', 'NA', 'K', 'CA', 'CO', 'CL', 'MO']

for idx, row in check.iterrows():
    pdbcode = row['PDB Code']
    uniprot = row['Uniprot Entry']
    
    # download raw pdb file
    pdb.download_pdb(pdbcode)
    
    # access the file
    read_file_path = os.path.join(pdb.raw_dir, "%s.pdb"%pdbcode)
    read_file = open(read_file_path)
    
    for line in read_file:
        if (line[:6] == 'HETATM') and ("KCX" in line):
            if ("CX" in line.split('KCX')[0]) or ("OQ1" in line) or ("OQ2" in line):
                continue
            else:
                words = line.split()
                
                # there are some special cases, so words[1] or words[2]
                if words[1] in list_of_metals or words[2] in list_of_metals:
                    print('Uniprot:', row['Uniprot Entry'], 'PDB:', pdbcode)
                    print(line)
                    
                    resid = line[:30].split()[-1]
                    # special cases
                    try:
                        resid = int(resid)
                    except:
                        resid = int(resid[1:])

                    # exclude this pdb file if this lysine is the one we want
                    if resid == PLMD[uniprot][0]:
                        check.loc[idx,'Label'] = 'Excluded'
    read_file.close()
```

```python
check = check[check['Label'] == 'Included']
```
