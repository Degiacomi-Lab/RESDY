# SELECTION OF REPRESENTATIVE DATA POINTS

### This Jupyter notebook serves as a guide for selecting representative data points from a large dataset.

```python
import pandas as pd
from preprocessing import Preprocessing
```

#### Load a dataset from a csv file:

```python
df = pd.read_csv('measures_cut_KingAllNegative_all.csv')

df
```

Select features of interest (the AEV will be selected automaticaly, so only scalar features need to be specified), then use the list as an argument for the class <code>Negetive_preprocessing</code>.

```python
features = ['das', 'depth', 'sasa']
NP = Preprocessing(df, features)
```

The dataset after being cleaned is stored in the attribute <code>data_full</code>. The columns 0,1,2,... are the components of the AEVs having at least one nonzero entry in all observations.

```python
NP.data_full
```

The clustering methods are distance-based, so normalisation is required. All selected columns are rescaled to have mean zero and standard deviation one.

```python
NP.normalise()
NP.data_normalised
```

Finally, we obtain the undersampled dataset. The observations returned may vary between runs due to the randomness of the clustering methods used.



#### Method:

First separate the dataset into two parts, the 'common' and the 'outliers', based on interquartile range (IQR). The sensitivity can be changed by fixing the argument <code>iqr_reject_range</code>: all data points having at least one feature lie in the rejection region are considered as outliers; this region is defined by $[25^{th} \text{ percentile} - \text{iqr_reject_range}*\text{IQR}, 75^{th} \text{ percentile} + \text{iqr_reject_range}*\text{IQR}]$.
        
+) Common part: use K-means to group data points into clusters, then from each cluster choose the point closest to cluster centre.
        
+) Outlier part: use DBSCAN algorithm to find small clusters in the set of outliers, then from each cluster choose the point closest to cluster centre. The sensitivity can be tuned by changing the argument <code>outlier_cluster_radius</code>.

```python
NP.undersampling(features, 800, N_init = 100, iqr_reject_range=1.5, outlier_cluster_radius=0.6)
```

```python
NP.undersampled_data
```

### The scatterplots below are for comparisons between datasets

```python
import matplotlib.pyplot as plt

fig = plt.figure(figsize=(18,10))
ax = fig.add_subplot(1,3,1, projection='3d')
ax.scatter(NP.data['depth'], NP.data['das'], NP.data['sasa'], s=3, alpha=0.5)
ax.set_xlabel('depth')
ax.set_ylabel('das')
ax.set_zlabel('sasa');
ax.set_title('Full dataset');

ax1 = fig.add_subplot(1,3,2, projection='3d')
ax1.scatter(NP.undersampled_data['depth'], NP.undersampled_data['das'], NP.undersampled_data['sasa'], s=3, alpha=0.5)
ax1.set_xlabel('depth')
ax1.set_ylabel('das')
ax1.set_zlabel('sasa');
ax1.set_title('Undersampled dataset');

ax2 = fig.add_subplot(1,3,3, projection='3d')
ax2.scatter(NP.undersampled_data_common['depth'], NP.undersampled_data_common['das'], NP.undersampled_data_common['sasa'], s=3, alpha=0.5)
ax2.scatter(NP.undersampled_data_outlier['depth'], NP.undersampled_data_outlier['das'], NP.undersampled_data_outlier['sasa'], s=3, alpha=0.5, c='r')
ax2.set_xlabel('depth')
ax2.set_ylabel('das')
ax2.set_zlabel('sasa');
ax2.set_title('Undersampled dataset - Common (blue) vs Outliers (red)');
```
